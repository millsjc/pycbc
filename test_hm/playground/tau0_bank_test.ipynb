```python
%pylab inline
from pycbc.conversions import tau0_from_mass1_mass2
import h5py
```

```python
inputfile = "../bank/full-correct-tau-xhm-spinning-2filter-listharms.hdf"
with h5py.File(inputfile, "r") as f:
    params = {k: array(v) for k,v in f.items()}
```

```python
m1_min, m2_min = min(params["mass1"]), min(params["mass2"])
m1_max, m2_max = max(params["mass1"]), max(params["mass2"])
m1, m2 = linspace(m1_min,m1_max), linspace(m2_min,m2_max)
m1,m2 = meshgrid(m1, m2)
t0 = tau0_from_mass1_mass2(m1, m2, 20)
contourf(m1, m2, t0)
plot(params["mass1"], params["mass2"], "rx", alpha=0.05)
colorbar()
```
