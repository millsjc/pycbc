We investigate a simpler algorithm for computing the possible combinations of interferometers, rather than having a pre-computed index array which is then used to select the different possible combinations and find the maximum.

Instead, we use a sliding window approach where the maximum within each of the possible differences is computed in the second array, in the third tertiary detector and then you can do this without creating a new array, so just using a view which is efficient.

It's more memory efficient and hopefully also in terms of time.

```python
import numpy as np

# Example array
arr = np.array([1, 3, 2, 5, 8, 1, 2, 0, 6, 7])

# Window size
window_size = 3

# Create sliding windows
windows = np.lib.stride_tricks.sliding_window_view(arr, window_shape=window_size)

# Compute max along the last axis
moving_max = windows.max(axis=1)
print(moving_max)
```
