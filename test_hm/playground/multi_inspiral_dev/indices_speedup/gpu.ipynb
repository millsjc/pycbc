```python
import torch
import time

# size = 50000
# timeseries = arange(0,5000,100)
n_comb = int(1e3)
dtype = torch.float32

# Create two random matrices

a_cpu = torch.randn(n_comb, 3, dtype=dtype)
b_cpu = torch.randn(n_comb, 3, dtype=dtype)

# Perform the matrix multiplication on the CPU
start_time = time.time()
a = torch.matmul(a_cpu, b_cpu.T)
print(a)
cpu_time = time.time() - start_time
print(f'CPU time: {cpu_time:.6f} seconds')

# Move the matrices to the GPU
# device = 'mps:0'
a_gpu = a_cpu.to('mps')
b_gpu = b_cpu.to('mps')

# Perform the matrix multiplication on the GPU
start_time = time.time()
a = torch.matmul(a_gpu, b_gpu.T)
print(a)
gpu_time = time.time() - start_time
print(f'GPU time: {gpu_time:.6f} seconds')

# Compare the times
print(f'The GPU was {cpu_time / gpu_time:.2f} times faster than the CPU')
```

```python
import torch
import time
import numpy as np

# Create some example pycbc time series data
data1 = np.array(np.random.rand(5000), dtype=np.float32)
data2 = np.array(np.random.rand(5000), dtype=np.float32)
data3 = np.array(np.random.rand(5000), dtype=np.float32)

data_list = [torch.from_numpy(data1), 
             torch.from_numpy(data2), 
             torch.from_numpy(data3)]

# Generate all combinations of indices. (this step might be unrealistic for large N)
comb_indices = torch.tensor(np.random.randint(0, 5000-1, size=((int(3*1e6)), 3)))

def compute_RSS(data_list, indices, device):
    data_list = [data.to(device) for data in data_list]
    indices = indices.to(device)
    RSS = (data_list[0][indices[:, 0]]**2 +
           data_list[1][indices[:, 1]]**2 +
           data_list[2][indices[:, 2]]**2).sqrt()
    idx = indices[RSS > 7]
    return RSS, idx

# Test on CPU
start_time = time.time()
RSS, idx_cpu = compute_RSS(data_list, comb_indices, device=torch.device('cpu'))
# results_cpu = comb_indices[RSS > 7]
end_time = time.time()
print(f"CPU computation time: {end_time - start_time} seconds")

# Test on GPU
start_time = time.time()
RSS, idx_gpu = compute_RSS(data_list, comb_indices, device=torch.device('mps'))
# results_gpu = comb_indices[RSS > 7]
end_time = time.time()
print(f"GPU computation time: {end_time - start_time} seconds")
```
